```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm_model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")#"gemini-2.5-flash")
# llm_model = ChatGoogleGenerativeAI(model="gemini-3-flash-preview")
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import PydanticOutputParser, JsonOutputParser,StrOutputParser
```

```python
# Description chain

description_txt = "Generate 2 line description for given topic {topic}"
description_prompt = ChatPromptTemplate.from_template(description_txt)

description_chain = (
    description_prompt
    |
    llm_model
    |
    StrOutputParser()
)
```

```python
# Pros chain

pro_txt = "Generate 3 bullet points describing pros about the topic {topic}"
pro_prompt = ChatPromptTemplate.from_template(pro_txt)

pro_chain = (
    pro_prompt
    |
    llm_model
    |
    StrOutputParser()
)
```

```python
# Cons Chain

cons_txt = "Generate 3 bullet points describing cons about the topic {topic}"
cons_prompt = ChatPromptTemplate.from_template(cons_txt)

cons_chain = (
    cons_prompt
    |
    llm_model
    |
    StrOutputParser()
)
```

```python
from operator import itemgetter
from langchain_core.runnables import RunnableParallel
```

```python
branch_chain =(
    RunnableParallel(
        topic=itemgetter('topic'),
        description=description_chain,
        pros=pro_chain,
        cons=cons_chain,
        
    )
)
# running multiple chains in parallely on same topic using runnableparallel
```

```python
branch_chain.invoke({'topic':'stock market'})
```

```python
# merge report

merge_report = """

    Create a report about topic {topic} with the following information:

    Description:
    {description}

    Pros:
    {pros}

    Cons:
    {cons}

    Report should be in the following format:

    Topic: <Name of the topic>

    Description: <Description of the topic>

    Pros and Cons:
    <Table with 2 columns with pros and cons side by side>
    

"""


merge_prompt = ChatPromptTemplate.from_template(merge_report)

merge_chain = (
    merge_prompt
    |
    llm_model
    |
    StrOutputParser()
)
```

```python
final_chain = (
    branch_chain
    |
    merge_chain
)

response = final_chain.invoke({'topic':"real state"})
```

```python
from IPython.display import Markdown,display

display(Markdown(response))
```

```python
classifier_txt = """
    Given the user instruction below for analysing customer review,
    classify it as only one of the following categories:
    -summarise
    -sentiment
    -email

    do not respond other then one word.
    Instructions:
    {instruction}


"""

classifier_prompt = ChatPromptTemplate.from_template(classifier_txt)

classifier_chain = (
    classifier_prompt
    |
    llm_model
    |
    StrOutputParser()
)
```

```python
summary_txt = """
    Act as a customer review analyst, given the following customer review
    generate a short summary (max 2 lines) of the review

    Customer Review:
    {review}
    """

summary_prompt = ChatPromptTemplate.from_template(summary_txt)

summary_chain = (
    summary_prompt
    |
    llm_model
    |
    StrOutputParser()
)
```

```python
sentiment_txt = """
    Act as a customer review analyst, given the following customer review
    find out the sentiment of the review.
    The sentiment could be positive, negative or neutral.
    return the result as a single word

    Customer Review:
    {review}
    """

sentiment_prompt = ChatPromptTemplate.from_template(sentiment_txt)

sentiment_chain = (
    sentiment_prompt
    |
    llm_model
    |
    StrOutputParser()
)
```

```python
email_txt = """
    Act as a customer review analyst, given the following customer review
    generate an email to the customer based on the following condition
    -if sentiment is positive and neutral thank them
    -if sentiment is negative apolizise to them

    Customer Review:
    {review}

    Sentiment:
    {sentiment}
    """

email_prompt = ChatPromptTemplate.from_template(email_txt)

email_chain = (
    email_prompt
    |
    llm_model
    |
    StrOutputParser()
)
```

```python
def default_answer(query):
    return "instructions are not defined inthe intents"
```

```python
from langchain_core.runnables import RunnableBranch

branch = RunnableBranch(
    (lambda x: "summarise" in x['topic'].lower(), summary_chain),
    (lambda x: "sentiment" in x['topic'].lower(), sentiment_chain),
    (lambda x: "email" in x['topic'].lower(), email_chain),
    default_answer
)
# running single chain based on key
```

```python
full_chain = (
    {
        "topic": classifier_chain,
        "instruction": lambda input_prompt:input_prompt.get('instruction'),
        "review": lambda input_prompt:input_prompt.get('review'),
        "sentiment": lambda input_prompt:input_prompt.get('sentiment'),
    }
    |
    branch
)
```

```python
sample_review = """
    I recently purchased this table and I could not be happier with my choice fromthe moment I unboxed it, I was Impressed by the sleek design and
    high quality craftsmanship. The lamp has a modern yet timeless asthetic that fits perfectly with my home decor.
"""
```

```python
summary = full_chain.invoke({
    "instruction":"Generate a summary for the given review",
    "review":sample_review
})

print(summary)
```

```python
sentiment = full_chain.invoke({
    "instruction":"Find out the sentiment of the review",
    "review":sample_review
})

print(sentiment)
```

```python
email = full_chain.invoke({
    "instruction":"write an email for the given customer review",
    "review":sample_review,
    "sentiment":sentiment
})

print(email)
```
