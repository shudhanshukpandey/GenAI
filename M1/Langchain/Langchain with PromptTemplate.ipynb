```python
# from langchain.prompts import PromptTemplate
from langchain_core.prompts import PromptTemplate
```

```python
prompt_txt = "Explain Gen AI in 3 bullet points"
prompt = PromptTemplate.from_template(prompt_txt)
print(prompt)
```

```python

from langchain_google_genai import ChatGoogleGenerativeAI
llm_model = ChatGoogleGenerativeAI(model="gemini-3-flash-preview")#"gemini-2.5-flash")

chain = (
    prompt
    |
    llm_model
)

response = chain.invoke({"topic":"France"})
response.content
```

```python
prompt_txt = "Explain to me briefly about {topic} in {language}"

prompt = PromptTemplate.from_template(prompt_txt)

chain = (
    prompt
    |
    llm_model
)
```

```python
inputs = [
    {"topic": "AI", "language": "hindi"},
    {"topic": "GenAI", "language": "english"}
]
```

```python
response = chain.batch(inputs)

for val in response:
    print(val.content)
    print("*"*20)
```
